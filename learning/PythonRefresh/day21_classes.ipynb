```python
#Exercise 1
class Statistics:
    def __init__(self, data):
        self.data = data

    def count(self):
        return len(self.data)

    def sum(self):
        return sum(self.data)

    def min(self):
        return min(self.data)

    def max(self):
        return max(self.data)

    def range(self):
        return self.max() - self.min()

    def mean(self):
        return self.sum()/self.count()

    def median(self):
        s = sorted(self.data)
        n = len(s)      
        mid = n // 2               
        if n % 2 == 1:             
            return s[mid]
        else:                      
            return (s[mid - 1] + s[mid]) / 2

    def mode(self):
        counts = {}
        for x in self.data:
            counts[x] = counts.get(x, 0) + 1

        best_value = None
        best_count = 0
        for value, count in counts.items():
            if count > best_count:
                best_value = value
                best_count = count

        return {'mode': best_value, 'count': best_count}

    def var(self):
        total = 0
        for x in self.data:
            total+=(x - self.mean())**2
        return round(total/self.count(),1)

    def std(self):
        return round(self.var()**0.5, 1)

    def freq_dist(self):
        counts = {}
        for x in self.data:
            counts[x] = counts.get(x, 0) + 1

        dist = []
        for value, count in counts.items():
            percentage = count / self.count() *100
            dist.append((percentage, value))

        return sorted(dist, reverse=True)
        

ages = [31, 26, 34, 37, 27, 26, 32, 32, 26, 27, 27, 24, 32, 33, 27, 25, 26, 38, 37, 31, 34, 24, 33, 29, 26]

data = Statistics(ages)

print('Count:', data.count()) # 25
print('Sum: ', data.sum()) # 744
print('Min: ', data.min()) # 24
print('Max: ', data.max()) # 38
print('Range: ', data.range()) # 14
print('Mean: ', data.mean()) # 30
print('Median: ', data.median()) # 29
print('Mode: ', data.mode()) # {'mode': 26, 'count': 5}
print('Standard Deviation: ', data.std()) # 4.2
print('Variance: ', data.var()) # 17.5
print('Frequency Distribution: ', data.freq_dist()) # [(20.0, 26), (16.0, 27), (12.0, 32), (8.0, 37), (8.0, 34), (8.0, 33), (8.0, 31), (8.0, 24), (4.0, 38), (4.0, 29), (4.0, 25)]
```

```python
#Exercise 2
#Create a class called PersonAccount. It has firstname, lastname, incomes, expenses properties and it has total_income, total_expense, account_info, add_income, add_expense and account_balance methods. Incomes is a set of incomes and its description. The same goes for expenses.

class PersonAccount:
    def __init__(self, firstname, lastname, incomes, expenses):
        self.firstname = firstname
        self.lastname = lastname
        self.incomes = incomes
        self.expenses = expenses

    def total_income(self):
        return sum(self.incomes.values())

    def total_expenses(self):
        return sum(self.expenses.values())

    def account_balance(self):
        return self.total_income()-self.total_expenses()

    def account_info(self):
        return f"{self.firstname} {self.lastname}'s Balance: {self.account_balance()}"

    def add_income(self, description, amount):
        self.incomes[description] = amount

    def add_expense(self, description, amount):
        self.expenses[description] = amount


p=PersonAccount("Adrien", "Jean", {"Salary": 2000, "Rental Income": 1500}, {"Bills": 150, "Mortgage": 1500})
print(p.firstname)
print(p.total_income())
print(p.total_expenses())
print(p.account_info())
p.add_income("Bonus", 500)
p.add_expense("Car Repair", 100)
print(p.total_income())
print(p.total_expenses())       # should go from 3500 to 4000
print(p.account_info())      # balance should go up by 500
```
